# Scientific-Conclusion Stability under Catalog-Source Substitution

**Protocol 2.4: exact finite-source certificates for exoplanet-derived quantities and
transit-only operational rankings**

This notebook compares two frozen, analysis-ready exoplanet datasets without treating
either source as ground truth. It asks whether derived physical conclusions remain stable
when source values are substituted. The primary endpoint is the geometric planet-to-star
area ratio. A separate operational endpoint ranks only systems classified as transit-related
in both datasets.

## Estimands and interpretation

For positive source values $x_{j,A}$ and $x_{j,B}$ and a monomial observable
$f(x)=C\prod_j x_j^{\alpha_j}$, this notebook distinguishes:

1. **complete-row substitution**, comparing $f_A$ with $f_B$;
2. **coordinate-wise source-factorial substitution**, evaluating every A/B input vertex;
3. **decision stability**, determining whether threshold or top-$k$ decisions are invariant.

These are exact properties of the two supplied artifacts. The coordinate-wise factorial
interval constitutes a **finite-source sensitivity envelope**---the range of derived
values produced by exhaustive A/B substitution at each coordinate. It is not a physical
uncertainty interval, not a confidence or credible interval for the unknown true planet,
and does not identify which source is more accurate.


## Exact-bound proposition

Let $\ell_j=\min(\log x_{j,A},\log x_{j,B})$ and
$u_j=\max(\log x_{j,A},\log x_{j,B})$. For a positive monomial,

$$
\log f_{\min}=\log C+\sum_{\alpha_j\ge0}\alpha_j\ell_j
  +\sum_{\alpha_j<0}\alpha_j u_j,
$$
$$
\log f_{\max}=\log C+\sum_{\alpha_j\ge0}\alpha_j u_j
  +\sum_{\alpha_j<0}\alpha_j\ell_j.
$$

Therefore
$$
\log(f_{\max}/f_{\min})=
\sum_j |\alpha_j|\,|\log(x_{j,A}/x_{j,B})|.
$$

The implementation verifies this result by enumerating every source vertex for every
eligible row and every observable.

## Corrigendum: K2-136 d unit correction (protocol 2.4.0)

The raw Source A snapshot (`data/raw/kepler.csv`, sha256 `74b9c7ae...`) contained a
unit error for **K2-136 d**: its radius was entered as `1.450 R_Jup` although the underlying
source value is `1.450 R_earth` (Mann et al. 2018). Interpreted as Jupiter radii it produced
the largest substitution width in the raw baseline (factorial width ~113.48, rank_a 2,
rank_b 1380, rank displacement 1378) and was the only `unit_error` forensic finding.

Corrected snapshot (sha256 `c945f9a5...`):
- radius: `1.450 R_Jup` -> `0.1293603354 R_Jup` (= `1.450 R_earth` / R_JUPITER_TO_R_EARTH=11.209);
- radius error bars: `0.080/0.110 R_Jup` -> `0.0071371220/0.0098135427 R_Jup`
  (= `0.080/0.110 R_earth`, matching Mann et al. 2018 `1.45_{-0.08}^{+0.11} R_earth`);
- every other row and column is byte-identical to the raw snapshot.

After correction the primary maximum factorial width falls from ~113.48 to **13.57**
(`hats-12 b`, rank displacement 85). K2-136 d is no longer an extreme case and is absent
from the forensic table. The operational robust pool holds **44** concordant transits
(13 guaranteed, 31 possible top-25 members), and all 113 priority-0 cases received documented review. Of these, 105 were
classified as confirmed source differences and eight as confirmed agreements. This corrigendum is
recorded under `source_a.derivative_status` and `source_a.preprocessing_history` in
`source_provenance.json`.


## Discussion: K2-136 d — historical versus composite radius

After the unit correction (see corrigendum above), the two frozen sources still differ on
the radius of **K2-136 d**: `1.450 R_earth` (Source A, historical, Mann et al. 2018) versus
`1.565 R_earth` (Source B, NASA composite). The relative difference is about **7.9%**
(`1.565 / 1.450 - 1`).

This is a genuine catalog-version and measurement-epoch difference, not a second unit error.
The two sources differ not only in catalog curation policies but also in publication epoch:
Source A extends to approximately February 2018, whereas Source B is frozen in March 2026.
Observed discrepancies therefore combine curation decisions and temporal updates:

- Source A preserves the historical published value `1.45_{-0.08}^{+0.11} R_earth`
  (Mann et al. 2018);
- Source B reports the NASA Exoplanet Archive composite value `1.565 R_earth`.

The protocol 2.4.0 corrigendum corrected only the Source A *unit encoding*
(`1.450 R_Jup` -> `0.1293603354 R_Jup`) without altering the historical value, so the
remaining discrepancy is retained as a `confirmed_source_difference` and requires no second
correction. K2-136 d remains concordant under the operational transit-evidence criterion;
its radius difference contributes to the primary geometric-width distribution and is
subject to the same forensic case-selection rules as every other matched system.

**Source-date confounding:** the comparison confounds catalog curation differences with the ~8-year information-epoch gap (Source A ~February 2018, Source B March 2026). Disentangling these two effects would require access to same-epoch catalog versions, which is outside the scope of this finite-source analysis. Readers should interpret all observed discrepancies as combining both sources of variation.


## Discussion: aggregate agreement versus tail rank displacement

The two-source ranking of the `N = 2,561` concordant-transit geometric signal values shows
high aggregate agreement: Kendall tau `0.9765`, Spearman rho `0.9967`, median rank
displacement `5.0`, 95th percentile `77.0`, 99th percentile `250.4`.

The tail is far larger than the aggregate suggests: the **maximum rank displacement is
`1,115`** positions. Within the operational top-25 pool the largest displacement is `243`
(`hat-p-67 b`), so the extreme `1,115`-place movement occurs below the top-25 boundary. The
top-25 sets overlap in `21 / 29` members (Jaccard `0.724`), giving a robust pool of `44`
(`13` guaranteed, `31` possible).

Interpretation: aggregate agreement statistics compress the extreme cases that the forensic
case-selection procedure is designed to surface. Any ranking-based conclusion must report
both the aggregate coefficients and the tail displacements, and extreme cases must cite the
underlying `confirmed_source_difference` review record rather than a blanket agreement
claim.

```python
def sha256_file(path: Path) -> str:
    """sha256_file()."""
    if not path.is_file():
        raise FileNotFoundError(path)
    digest = hashlib.sha256()
    with path.open("rb") as handle:
        for block in iter(lambda: handle.read(1 << 20), b""):
            digest.update(block)
    return digest.hexdigest()


def json_default(value: Any) -> Any:
    """json_default()."""
    if isinstance(value, np.bool_):
        return bool(value)
    if isinstance(value, np.integer):
        return int(value)
    if isinstance(value, np.floating):
        return float(value)
    if isinstance(value, Path):
        return value.as_posix()
    if isinstance(value, (tuple, set)):
        return list(value)
    raise TypeError(f"Unsupported JSON value: {type(value)!r}")


def sanitize_json(value: Any) -> Any:
    """sanitize_json()."""
    if isinstance(value, Mapping):
        return {str(key): sanitize_json(item) for key, item in value.items()}
    if isinstance(value, (list, tuple, set)):
        return [sanitize_json(item) for item in value]
    if isinstance(value, (float, np.floating)) and not math.isfinite(float(value)):
        return None
    if isinstance(value, np.bool_):
        return bool(value)
    if isinstance(value, np.integer):
        return int(value)
    if isinstance(value, np.floating):
        return float(value)
    return value


def prepare_output_directories(paths: AnalysisPaths, allow_replacement: bool) -> None:
    """prepare_output_directories()."""
    paths.validate()
    if paths.output_root.exists() and any(paths.output_root.rglob("*")) and not allow_replacement:
        raise FileExistsError(
            f"Definitive output directory is not empty: {paths.output_root}. "
            "Archive or remove it before a clean publication run."
        )
    for directory in (paths.artifact_dir, paths.figure_dir, paths.table_dir, paths.staging_dir):
        directory.mkdir(parents=True, exist_ok=True)


def atomic_json(payload: Mapping[str, Any], path: Path) -> None:
    """atomic_json()."""
    text = json.dumps(sanitize_json(payload), indent=2, sort_keys=True, allow_nan=False, default=json_default) + "\n"
    temporary = PATHS.staging_dir / f"{path.name}.tmp"
    temporary.write_text(text, encoding="utf-8", newline="\n")
    temporary.replace(path)


def atomic_csv(frame: pd.DataFrame, path: Path, *, gzip: bool = False) -> None:
    """atomic_csv()."""
    if frame.columns.duplicated().any():
        raise ValueError(f"Duplicate columns in export: {path}")
    temporary = PATHS.staging_dir / f"{path.name}.tmp"
    compression: Any = {"method": "gzip", "compresslevel": 9, "mtime": 0} if gzip else None
    frame.to_csv(temporary, index=False, compression=compression, encoding="utf-8", lineterminator="\n")
    temporary.replace(path)


def save_figure(fig: mpl.figure.Figure, stem: str) -> Path:
    """save_figure()."""
    path = PATHS.figure_dir / f"{stem}.{CONFIG.figure_format}"
    temporary = PATHS.staging_dir / f"{stem}.{CONFIG.figure_format}"
    fig.savefig(temporary, dpi=CONFIG.figure_dpi, bbox_inches="tight", metadata={"Title": stem, "Author": "Protocol 2.4 source-substitution study"})
    plt.close(fig)
    if not temporary.is_file() or temporary.stat().st_size == 0:
        raise RuntimeError(f"Figure export failed: {stem}")
    temporary.replace(path)
    reader = PdfReader(path)
    if len(reader.pages) != 1:
        raise RuntimeError(f"Figure must contain exactly one PDF page: {path}")
    media_box = reader.pages[0].mediabox
    if float(media_box.width) <= 0 or float(media_box.height) <= 0:
        raise RuntimeError(f"Figure has an invalid PDF media box: {path}")
    return path


def stable_sort(frame: pd.DataFrame, columns: Sequence[str]) -> pd.DataFrame:
    """stable_sort()."""
    return frame.sort_values(list(columns), kind="mergesort", na_position="last").reset_index(drop=True)


def load_optional_json(path: Path) -> dict[str, Any] | None:
    """load_optional_json()."""
    if not path.is_file():
        return None
    payload = json.loads(path.read_text(encoding="utf-8"))
    if not isinstance(payload, dict):
        raise ValueError(f"Optional JSON input must contain an object: {path}")
    return payload


def optional_file_record(path: Path) -> dict[str, Any] | None:
    """optional_file_record()."""
    if not path.is_file():
        return None
    return {"path": path.as_posix(), "sha256": sha256_file(path), "bytes": path.stat().st_size}


prepare_output_directories(PATHS, CONFIG.allow_output_replacement)
```

```python
def normalize_identifier(value: Any) -> str | None:
    """normalize_identifier()."""
    if value is None or pd.isna(value):
        return None
    text = unicodedata.normalize("NFKC", str(value)).strip().casefold()
    text = text.replace("−", "-").replace("–", "-").replace("—", "-")
    text = re.sub(r"\s*-\s*", "-", text)
    text = re.sub(r"\s+", " ", text)
    return text or None


def extract_planet_component(value: Any) -> str | None:
    """extract_planet_component()."""
    key = normalize_identifier(value)
    if key is None:
        return None
    match = re.search(r"(?:\s|\d)([b-z])$", key)
    return match.group(1) if match else None


def detection_rule(value: Any) -> DetectionRule:
    """detection_rule()."""
    key = normalize_identifier(value)
    if key is None:
        return DetectionRule("missing", "missing", "missing", "Source value is missing.")
    if key not in DETECTION_RULES:
        raise ValueError(f"Unreviewed discovery-method label: {value!r}")
    return DETECTION_RULES[key]


def standardize_detection_method(value: Any) -> str:
    """standardize_detection_method()."""
    return detection_rule(value).discovery_family


def standardize_transit_evidence(value: Any) -> str:
    """standardize_transit_evidence()."""
    return detection_rule(value).transit_evidence


def transit_concordance(evidence_a: pd.Series, evidence_b: pd.Series) -> pd.Series:
    """transit_concordance()."""
    a = evidence_a.eq("confirmed")
    b = evidence_b.eq("confirmed")
    return pd.Series(np.select(
        [a & b, a & ~b, ~a & b, ~a & ~b],
        ["both_transit", "a_transit_only", "b_transit_only", "neither_confirmed_transit"],
        default="unresolved",
    ), index=evidence_a.index, dtype="string")
```

```python
def load_csv(path: Path) -> pd.DataFrame:
    """load_csv()."""
    if not path.is_file():
        raise FileNotFoundError(f"Required frozen input is missing: {path}")
    frame = pd.read_csv(path, low_memory=False)
    if frame.empty:
        raise ValueError(f"Input is empty: {path}")
    if frame.columns.duplicated().any():
        raise ValueError(f"Input contains duplicate column names: {path}")
    return frame


def audit_raw_schema(frame: pd.DataFrame, schema: DatasetSchema, stage: str) -> ValidationReport:
    """audit_raw_schema()."""
    report = ValidationReport(stage)
    required = {schema.planet_name_column, schema.host_name_column, schema.discovery_method_column}
    required.update(spec.source_name for spec in schema.columns)
    missing = sorted(required - set(frame.columns))
    if missing:
        report.add("blocking", "missing_columns", f"Missing declared columns: {missing}", len(missing))
    report.metrics.update(rows=len(frame), columns=len(frame.columns), duplicate_rows=int(frame.duplicated().sum()))
    if schema.planet_name_column in frame and frame[schema.planet_name_column].isna().any():
        report.add("major", "missing_planet_names", "Planet-name column contains missing values.", int(frame[schema.planet_name_column].isna().sum()))
    return report


def convert_numeric(series: pd.Series, spec: ColumnSpec) -> tuple[pd.Series, pd.Series]:
    """convert_numeric()."""
    numeric = pd.to_numeric(series, errors="coerce")
    invalid = series.notna() & numeric.isna()
    if spec.positive:
        invalid |= numeric.notna() & (numeric <= 0)
    if spec.minimum is not None:
        invalid |= numeric.notna() & (numeric < spec.minimum)
    if spec.maximum is not None:
        invalid |= numeric.notna() & (numeric > spec.maximum)
    canonical = (numeric * spec.conversion_factor).mask(invalid).astype(float)
    return canonical, invalid.astype(bool)


def build_canonical_table(raw: pd.DataFrame, schema: DatasetSchema, source_label: str) -> tuple[pd.DataFrame, ValidationReport]:
    """build_canonical_table()."""
    raw_check = audit_raw_schema(raw, schema, f"raw_schema_{source_label}")
    report = ValidationReport(f"canonicalization_{source_label}")
    if not raw_check.passed:
        report.issues.extend(raw_check.issues)
        return pd.DataFrame(), report
    output = pd.DataFrame(index=raw.index)
    output["source_label"] = source_label
    output["source_row"] = np.arange(len(raw), dtype=int)
    output["planet_name_source"] = raw[schema.planet_name_column].astype("string")
    output["host_name_source"] = raw[schema.host_name_column].astype("string")
    output["planet_key"] = raw[schema.planet_name_column].map(normalize_identifier).astype("string")
    output["host_key"] = raw[schema.host_name_column].map(normalize_identifier).astype("string")
    output["component"] = raw[schema.planet_name_column].map(extract_planet_component).astype("string")
    output["discovery_method_source"] = raw[schema.discovery_method_column].astype("string")
    output["discovery_method_class"] = raw[schema.discovery_method_column].map(standardize_detection_method).astype("string")
    output["transit_evidence"] = raw[schema.discovery_method_column].map(standardize_transit_evidence).astype("string")
    for spec in schema.columns:
        values, invalid = convert_numeric(raw[spec.source_name], spec)
        output[spec.canonical_name] = values
        output[f"{spec.canonical_name}_invalid"] = invalid
        if invalid.any():
            report.add("minor", "invalid_numeric", f"{spec.source_name}: invalid values set to missing.", int(invalid.sum()), column=spec.source_name)
    output["duplicate_planet_key"] = output["planet_key"].duplicated(keep=False) & output["planet_key"].notna()
    report.metrics.update(rows=len(output), columns=len(output.columns))
    report.metrics["missing_planet_key"] = int(output["planet_key"].isna().sum())
    report.metrics["duplicate_planet_key_rows"] = int(output["duplicate_planet_key"].sum())
    return output.reset_index(drop=True), report


def report_table(report: ValidationReport) -> pd.DataFrame:
    """report_table()."""
    return pd.DataFrame({
        "stage": [report.stage], "status": [report.status], "computational_pass": [report.passed],
        "major_count": [report.major_count], "blocking_count": [report.blocking_count],
        "issues": [" | ".join(f"{x.severity}:{x.code}:{x.message}" for x in report.issues)],
        **{key: [value] for key, value in report.metrics.items()},
    })
```

```python
def generate_name_candidates(canon_a: pd.DataFrame, canon_b: pd.DataFrame) -> pd.DataFrame:
    """generate_name_candidates()."""
    left = canon_a[canon_a["planet_key"].notna()].copy()
    right = canon_b[canon_b["planet_key"].notna()].copy()
    counts_a = left["planet_key"].value_counts()
    counts_b = right["planet_key"].value_counts()
    left["key_count_a"] = left["planet_key"].map(counts_a)
    right["key_count_b"] = right["planet_key"].map(counts_b)
    return left.merge(right, on="planet_key", how="inner", suffixes=("_a", "_b"), validate="many_to_many")


def add_matching_diagnostics(candidates: pd.DataFrame) -> pd.DataFrame:
    """add_matching_diagnostics()."""
    output = candidates.copy()
    pa = pd.to_numeric(output["period_days_a"], errors="coerce")
    pb = pd.to_numeric(output["period_days_b"], errors="coerce")
    available = pa.notna() & pb.notna() & (pa > 0) & (pb > 0)
    output["period_available_both"] = available
    output["period_signed_log_ratio"] = np.nan
    output["period_absolute_log_ratio"] = np.nan
    output["period_relative_difference"] = np.nan
    with np.errstate(divide="raise", invalid="raise", over="raise"):
        signed = np.log(pa[available] / pb[available])
    output.loc[available, "period_signed_log_ratio"] = signed
    output.loc[available, "period_absolute_log_ratio"] = np.abs(signed)
    output.loc[available, "period_relative_difference"] = np.abs(pa[available] - pb[available]) / np.minimum(pa[available], pb[available])
    output["host_key_concordant"] = output["host_key_a"].eq(output["host_key_b"])
    output["component_concordant"] = output["component_a"].eq(output["component_b"]) | output["component_a"].isna() | output["component_b"].isna()
    output["transit_concordance"] = transit_concordance(output["transit_evidence_a"], output["transit_evidence_b"])
    return output


def assign_match_tiers(candidates: pd.DataFrame, config: AnalysisConfig) -> pd.DataFrame:
    """assign_match_tiers()."""
    output = candidates.copy()
    unique = output["key_count_a"].eq(1) & output["key_count_b"].eq(1)
    rel = output["period_relative_difference"]
    output["match_tier_auto"] = "rejected"
    output["match_reason_auto"] = "duplicate_identifier_or_incompatible_period"
    tier1 = unique & output["period_available_both"] & rel.le(config.tier1_period_relative_tolerance)
    tier2 = unique & (~output["period_available_both"] | (rel.gt(config.tier1_period_relative_tolerance) & rel.le(config.tier2_period_relative_tolerance)))
    output.loc[tier1, ["match_tier_auto", "match_reason_auto"]] = ["tier1", "unique_identifier_period_consistent"]
    output.loc[tier2, ["match_tier_auto", "match_reason_auto"]] = ["tier2", "unique_identifier_period_missing_or_looser"]
    return output


def load_match_review_decisions(path: Path) -> pd.DataFrame:
    """load_match_review_decisions()."""
    columns = [
        "source_row_a", "source_row_b", "match_tier", "reason", "reviewer", "review_date",
        "second_reviewer", "adjudication",
    ]
    if not path.is_file():
        return pd.DataFrame(columns=columns)
    decisions = pd.read_csv(path, dtype="string").fillna("")
    missing = set(columns) - set(decisions.columns)
    if missing:
        raise ValueError(f"Match review decisions are missing columns: {sorted(missing)}")
    if decisions.duplicated(["source_row_a", "source_row_b"]).any():
        raise ValueError("Match review decision keys must be unique.")
    for column in ("source_row_a", "source_row_b"):
        numeric = pd.to_numeric(decisions[column], errors="coerce")
        if numeric.isna().any() or (numeric % 1).ne(0).any():
            raise ValueError(f"{column} must contain integer source-row identifiers.")
        decisions[column] = numeric.astype(int)
    if not decisions["match_tier"].isin(["tier1", "tier2", "rejected"]).all():
        raise ValueError("Match review decisions contain an invalid tier.")
    dates = pd.to_datetime(decisions["review_date"], errors="coerce", utc=True)
    if dates.isna().any() or dates.gt(pd.Timestamp.now(tz="UTC")).any():
        raise ValueError("Match review dates must be valid and nonfuture.")
    reviewer_one = decisions["reviewer"].str.strip().str.casefold()
    reviewer_two = decisions["second_reviewer"].str.strip().str.casefold()
    if reviewer_one.eq("").any() or reviewer_two.eq("").any() or reviewer_one.eq(reviewer_two).any():
        raise ValueError("Match decisions require two distinct named reviewers.")
    if decisions["reason"].str.strip().eq("").any() or decisions["adjudication"].str.strip().eq("").any():
        raise ValueError("Match decisions require reasons and adjudications.")
    return decisions[columns].copy()


def apply_manual_decisions(candidates: pd.DataFrame, decisions: pd.DataFrame) -> pd.DataFrame:
    """apply_manual_decisions()."""
    output = candidates.copy()
    output["match_tier"] = output["match_tier_auto"]
    output["match_reason"] = output["match_reason_auto"]
    output["manual_review"] = False
    if decisions.empty:
        return output
    required = {"source_row_a", "source_row_b", "match_tier", "reason", "reviewer", "review_date", "second_reviewer", "adjudication"}
    if not required.issubset(decisions.columns):
        raise ValueError(f"Manual decisions require columns: {sorted(required)}")
    if decisions.duplicated(["source_row_a", "source_row_b"]).any():
        raise ValueError("Manual decision keys must be unique.")
    candidate_keys = set(map(tuple, output[["source_row_a", "source_row_b"]].to_numpy()))
    decision_keys = set(map(tuple, decisions[["source_row_a", "source_row_b"]].to_numpy()))
    unknown = decision_keys - candidate_keys
    if unknown:
        raise ValueError(f"Manual match decisions reference unknown candidates: {sorted(unknown)[:10]}")
    lookup = decisions.set_index(["source_row_a", "source_row_b"])
    for index, row in output.iterrows():
        key = (row["source_row_a"], row["source_row_b"])
        if key in lookup.index:
            decision = lookup.loc[key]
            output.at[index, "match_tier"] = decision["match_tier"]
            output.at[index, "match_reason"] = f"manual:{decision['reason']}"
            output.at[index, "manual_review"] = True
    return output


def finalize_matches(candidates: pd.DataFrame) -> tuple[pd.DataFrame, pd.DataFrame]:
    """finalize_matches()."""
    audit = stable_sort(candidates, ["planet_key", "source_row_a", "source_row_b"])
    accepted = audit[audit["match_tier"].isin(["tier1", "tier2"])].copy()
    if accepted.duplicated("source_row_a").any() or accepted.duplicated("source_row_b").any():
        raise RuntimeError("Accepted matches are not one-to-one.")
    return audit, accepted.reset_index(drop=True)
```

```python
def semantic_eligibility(frame: pd.DataFrame, observable: PhysicalObservable, suffix: str) -> pd.Series:
    """semantic_eligibility()."""
    mask = pd.Series(True, index=frame.index)
    for variable in observable.inputs:
        column = f"{variable}{suffix}"
        if column not in frame:
            return pd.Series(False, index=frame.index)
        values = pd.to_numeric(frame[column], errors="coerce")
        mask &= values.notna() & np.isfinite(values) & values.gt(0)
    return mask


def evaluate_monomial(frame: pd.DataFrame, observable: PhysicalObservable, suffix: str) -> pd.Series:
    """evaluate_monomial()."""
    eligible = semantic_eligibility(frame, observable, suffix)
    result = pd.Series(np.nan, index=frame.index, dtype=float)
    if eligible.any():
        log_value = np.full(int(eligible.sum()), math.log(observable.constant), dtype=float)
        with np.errstate(divide="raise", invalid="raise", over="raise"):
            for variable, exponent in zip(observable.inputs, observable.exponents):
                log_value += exponent * np.log(frame.loc[eligible, f"{variable}{suffix}"].to_numpy(float))
            result.loc[eligible] = np.exp(log_value)
    return result


def enumerate_vertices(row: pd.Series, observable: PhysicalObservable) -> np.ndarray:
    """enumerate_vertices()."""
    pairs = [(float(row[f"{name}_a"]), float(row[f"{name}_b"])) for name in observable.inputs]
    values: list[float] = []
    for vertex in itertools.product(*pairs):
        log_value = math.log(observable.constant)
        for value, exponent in zip(vertex, observable.exponents):
            log_value += exponent * math.log(value)
        values.append(math.exp(log_value))
    return np.asarray(values, dtype=float)


def exact_factorial_bounds(frame: pd.DataFrame, observable: PhysicalObservable) -> pd.DataFrame:
    """exact_factorial_bounds()."""
    eligible_a = semantic_eligibility(frame, observable, "_a")
    eligible_b = semantic_eligibility(frame, observable, "_b")
    eligible = eligible_a & eligible_b
    output = pd.DataFrame(index=frame.index)
    output["eligible_a"] = eligible_a
    output["eligible_b"] = eligible_b
    output["eligible_both"] = eligible
    output["source_value_a"] = evaluate_monomial(frame, observable, "_a")
    output["source_value_b"] = evaluate_monomial(frame, observable, "_b")
    output["complete_row_lower"] = np.minimum(output["source_value_a"], output["source_value_b"])
    output["complete_row_upper"] = np.maximum(output["source_value_a"], output["source_value_b"])
    output["complete_row_log_width"] = np.log(output["complete_row_upper"] / output["complete_row_lower"])
    output["factorial_lower"] = np.nan
    output["factorial_upper"] = np.nan
    output["factorial_log_width"] = np.nan
    if eligible.any():
        lower_log = np.full(int(eligible.sum()), math.log(observable.constant))
        upper_log = np.full(int(eligible.sum()), math.log(observable.constant))
        for variable, exponent in zip(observable.inputs, observable.exponents):
            la = np.log(frame.loc[eligible, f"{variable}_a"].to_numpy(float))
            lb = np.log(frame.loc[eligible, f"{variable}_b"].to_numpy(float))
            lo, hi = np.minimum(la, lb), np.maximum(la, lb)
            lower_log += exponent * (lo if exponent >= 0 else hi)
            upper_log += exponent * (hi if exponent >= 0 else lo)
            contribution = pd.Series(np.nan, index=frame.index, dtype=float)
            contribution.loc[eligible] = abs(exponent) * np.abs(la - lb)
            output[f"contribution_{variable}"] = contribution
        output.loc[eligible, "factorial_lower"] = np.exp(lower_log)
        output.loc[eligible, "factorial_upper"] = np.exp(upper_log)
        output.loc[eligible, "factorial_log_width"] = upper_log - lower_log
    output["complete_row_width"] = np.exp(output["complete_row_log_width"])
    output["factorial_width"] = np.exp(output["factorial_log_width"])
    output["factorial_inflation"] = output["factorial_width"] / output["complete_row_width"]
    output["signed_log_ratio"] = np.log(output["source_value_a"] / output["source_value_b"])
    output["absolute_log_ratio"] = output["signed_log_ratio"].abs()
    return output


def build_certificates(matches: pd.DataFrame, observable: PhysicalObservable) -> pd.DataFrame:
    """build_certificates()."""
    bounds = exact_factorial_bounds(matches, observable)
    identity_columns = [
        "planet_key", "planet_name_source_a", "planet_name_source_b", "host_key_a", "host_key_b",
        "source_row_a", "source_row_b", "match_tier", "match_reason", "manual_review",
        "period_relative_difference", "discovery_method_class_a", "discovery_method_class_b",
        "transit_evidence_a", "transit_evidence_b", "transit_concordance",
    ]
    for variable in observable.inputs:
        identity_columns.extend([f"{variable}_a", f"{variable}_b"])
    output = pd.concat([matches[identity_columns].copy(), bounds], axis=1)
    output["observable"] = observable.name
    output["observable_unit"] = observable.unit
    return output


def verify_certificates(certificates: pd.DataFrame, matches: pd.DataFrame, observable: PhysicalObservable, tolerance: float, stage: str | None = None) -> ValidationReport:
    """verify_certificates()."""
    report = ValidationReport(stage or f"certificate_verification:{observable.name}")
    eligible_indices = certificates.index[certificates["eligible_both"]]
    failures = 0
    contribution_columns = [f"contribution_{name}" for name in observable.inputs]
    for index in eligible_indices:
        vertices = enumerate_vertices(matches.loc[index], observable)
        expected = certificates.loc[index, ["factorial_lower", "factorial_upper"]].to_numpy(float)
        if not np.allclose(expected, [vertices.min(), vertices.max()], rtol=tolerance, atol=0):
            failures += 1
    if failures:
        report.add("blocking", "vertex_mismatch", "Analytical and enumerated bounds differ.", failures)
    eligible = certificates.loc[eligible_indices]
    inside = eligible["source_value_a"].between(eligible["factorial_lower"], eligible["factorial_upper"], inclusive="both") & eligible["source_value_b"].between(eligible["factorial_lower"], eligible["factorial_upper"], inclusive="both")
    if not inside.all():
        report.add("blocking", "source_outside_bounds", "A complete-source value lies outside its factorial bounds.", int((~inside).sum()))
    contribution_sum = eligible[contribution_columns].sum(axis=1)
    if not np.allclose(contribution_sum, eligible["factorial_log_width"], rtol=tolerance, atol=tolerance):
        report.add("blocking", "contribution_identity", "Input contributions do not sum to log width.")
    finite_columns = ["source_value_a", "source_value_b", "factorial_lower", "factorial_upper", "factorial_width"]
    if not np.isfinite(eligible[finite_columns].to_numpy(float)).all():
        report.add("blocking", "nonfinite_values", "Eligible certificates contain non-finite values.")
    if (eligible["factorial_lower"] > eligible["factorial_upper"]).any() or (eligible["factorial_width"] < 1).any():
        report.add("blocking", "invalid_order", "Certificate bound ordering failed.")
    if (eligible["factorial_width"] + tolerance < eligible["complete_row_width"]).any():
        report.add("blocking", "factorial_nesting", "Factorial width is smaller than complete-row width.")
    report.metrics.update(eligible_rows=len(eligible_indices), enumerated_rows=len(eligible_indices), failures=failures)
    return report
```

```python
def add_source_ranks(certificates: pd.DataFrame) -> pd.DataFrame:
    """add_source_ranks()."""
    output = certificates.copy()
    for source in ("a", "b"):
        column = f"source_value_{source}"
        output[f"rank_{source}"] = output[column].rank(method="min", ascending=False, na_option="bottom").astype("Int64")
    output["rank_displacement"] = (output["rank_a"] - output["rank_b"]).abs().astype("Int64")
    return output


def robust_top_k(certificates: pd.DataFrame, k: int) -> pd.DataFrame:
    """robust_top_k()."""
    valid = certificates.dropna(subset=["factorial_lower", "factorial_upper"]).copy()
    lower = valid["factorial_lower"].to_numpy(float)
    upper = valid["factorial_upper"].to_numpy(float)
    sorted_upper = np.sort(upper)
    sorted_lower = np.sort(lower)
    could_outrank = len(valid) - np.searchsorted(sorted_upper, lower, side="right") - (upper > lower).astype(int)
    always_outrank = len(valid) - np.searchsorted(sorted_lower, upper, side="right")
    valid["worst_rank"] = 1 + np.maximum(could_outrank, 0)
    valid["best_rank"] = 1 + always_outrank
    valid["top_k_status"] = np.select(
        [valid["worst_rank"].le(k), valid["best_rank"].le(k)],
        ["guaranteed", "possible"], default="excluded",
    )
    valid["k"] = k
    return stable_sort(valid, ["worst_rank", "best_rank", "planet_key"])


def verify_ranking(ranking: pd.DataFrame, k: int, expected_n: int) -> ValidationReport:
    """verify_ranking()."""
    report = ValidationReport(f"robust_ranking:k={k}")
    valid_status = ranking["top_k_status"].isin(["guaranteed", "possible", "excluded"])
    if len(ranking) != expected_n:
        report.add("blocking", "ranking_size", "Ranking size differs from operational cohort.", len(ranking), expected=expected_n)
    if not valid_status.all():
        report.add("blocking", "invalid_status", "Unknown robust-ranking status.", int((~valid_status).sum()))
    if not ranking.loc[ranking["top_k_status"].eq("guaranteed"), "worst_rank"].le(k).all():
        report.add("blocking", "guarantee_condition", "Guaranteed item violates worst-rank condition.")
    if not ranking.loc[ranking["top_k_status"].eq("excluded"), "best_rank"].gt(k).all():
        report.add("blocking", "exclusion_condition", "Excluded item violates best-rank condition.")
    report.metrics.update(rows=len(ranking), guaranteed=int(ranking["top_k_status"].eq("guaranteed").sum()), possible=int(ranking["top_k_status"].eq("possible").sum()), excluded=int(ranking["top_k_status"].eq("excluded").sum()))
    return report


def ranking_agreement(certificates: pd.DataFrame, k: int) -> dict[str, Any]:
    """ranking_agreement()."""
    valid = certificates.dropna(subset=["source_value_a", "source_value_b"]).copy()
    top_a = set(valid.nlargest(k, "source_value_a")["planet_key"])
    top_b = set(valid.nlargest(k, "source_value_b")["planet_key"])
    intersection, union = top_a & top_b, top_a | top_b
    displacement = valid["rank_displacement"].astype(float)
    return {
        "n_ranking": len(valid),
        "spearman_rho": float(stats.spearmanr(valid["source_value_a"], valid["source_value_b"]).statistic),
        "kendall_tau": float(stats.kendalltau(valid["source_value_a"], valid["source_value_b"]).statistic),
        "top_k_intersection": len(intersection), "top_k_union": len(union),
        "top_k_jaccard": len(intersection) / len(union) if union else float("nan"),
        "median_rank_displacement": float(displacement.median()),
        "p95_rank_displacement": float(displacement.quantile(0.95)),
        "p99_rank_displacement": float(displacement.quantile(0.99)),
        "maximum_rank_displacement": int(displacement.max()),
    }
```

```python
def summarize_widths(certificates: pd.DataFrame, factors: Sequence[float]) -> dict[str, Any]:
    """summarize_widths()."""
    work = certificates.dropna(subset=["factorial_log_width"]).copy()
    factorial = work["factorial_width"]
    complete = work["complete_row_width"]
    summary: dict[str, Any] = {
        "n_primary": len(work),
        "median_factorial_width": float(factorial.median()),
        "q75_factorial_width": float(factorial.quantile(0.75)),
        "q90_factorial_width": float(factorial.quantile(0.90)),
        "q95_factorial_width": float(factorial.quantile(0.95)),
        "q99_factorial_width": float(factorial.quantile(0.99)),
        "maximum_factorial_width": float(factorial.max()),
        "median_complete_row_width": float(complete.median()),
        "median_factorial_inflation": float(work["factorial_inflation"].median()),
        "q90_factorial_inflation": float(work["factorial_inflation"].quantile(0.90)),
        "q95_factorial_inflation": float(work["factorial_inflation"].quantile(0.95)),
        "q99_factorial_inflation": float(work["factorial_inflation"].quantile(0.99)),
        "maximum_factorial_inflation": float(work["factorial_inflation"].max()),
    }
    for factor in factors:
        label = str(factor).replace(".", "p")
        summary[f"count_factorial_at_or_below_{label}"] = int(factorial.le(factor).sum())
        summary[f"fraction_factorial_at_or_below_{label}"] = float(factorial.le(factor).mean())
        summary[f"count_factorial_above_{label}"] = int(factorial.gt(factor).sum())
        summary[f"fraction_factorial_above_{label}"] = float(factorial.gt(factor).mean())
    for factor in (1.01, 1.05, 1.25, 2.0):
        label = str(factor).replace(".", "p")
        summary[f"fraction_inflation_above_{label}"] = float(work["factorial_inflation"].gt(factor).mean())
    return summary


def threshold_curve(certificates: pd.DataFrame, factors: Sequence[float]) -> pd.DataFrame:
    """threshold_curve()."""
    widths = certificates["factorial_width"].dropna()
    return pd.DataFrame({
        "factor": list(factors),
        "count_at_or_below": [int(widths.le(factor).sum()) for factor in factors],
        "fraction_at_or_below": [float(widths.le(factor).mean()) for factor in factors],
        "count_above": [int(widths.gt(factor).sum()) for factor in factors],
        "fraction_above": [float(widths.gt(factor).mean()) for factor in factors],
        "n_primary": len(widths),
    })


def k_curve(certificates: pd.DataFrame, values: Sequence[int]) -> pd.DataFrame:
    """k_curve()."""
    rows = []
    for k in values:
        ranking = robust_top_k(certificates, k)
        rows.append({
            "k": k,
            "guaranteed": int(ranking["top_k_status"].eq("guaranteed").sum()),
            "possible": int(ranking["top_k_status"].eq("possible").sum()),
            "excluded": int(ranking["top_k_status"].eq("excluded").sum()),
        })
    return pd.DataFrame(rows)


def paired_log_summary(a: pd.Series, b: pd.Series) -> dict[str, Any]:
    """paired_log_summary()."""
    valid = a.notna() & b.notna() & a.gt(0) & b.gt(0)
    logs = np.log(a[valid].to_numpy(float) / b[valid].to_numpy(float))
    if not len(logs):
        return {"n": 0}
    return {
        "n": len(logs), "median_signed_log_ratio": float(np.median(logs)),
        "median_absolute_log_ratio": float(np.median(np.abs(logs))),
        "q05_log_ratio": float(np.quantile(logs, 0.05)), "q95_log_ratio": float(np.quantile(logs, 0.95)),
    }


def host_composition_sensitivity(frame: pd.DataFrame, value_column: str, host_column: str, replicates: int, seed: int) -> pd.Series:
    """host_composition_sensitivity()."""
    work = frame.dropna(subset=[value_column]).copy()
    work[host_column] = work[host_column].fillna(work["planet_key"])
    groups = [group[value_column].to_numpy(float) for _, group in work.groupby(host_column, sort=True)]
    if not groups:
        raise ValueError("No host groups available for composition sensitivity.")
    generator = np.random.default_rng(seed)
    estimates = np.empty(replicates, dtype=float)
    for index in range(replicates):
        selected = generator.integers(0, len(groups), size=len(groups))
        estimates[index] = np.median(np.concatenate([groups[group] for group in selected]))
    return pd.Series(estimates, name="median_factorial_width")


def composition_interval(values: pd.Series, level: float) -> tuple[float, float]:
    """composition_interval()."""
    alpha = 1 - level
    return tuple(np.quantile(values.to_numpy(float), [alpha / 2, 1 - alpha / 2]).tolist())
```

```python
def select_forensic_cases(primary: pd.DataFrame, operational: pd.DataFrame, secondary: pd.DataFrame, config: AnalysisConfig) -> pd.DataFrame:
    """select_forensic_cases()."""
    operational_fields = operational[["planet_key", "best_rank", "worst_rank", "top_k_status"]].drop_duplicates("planet_key")
    primary_work = primary.drop(columns=["best_rank", "worst_rank", "top_k_status"], errors="ignore").merge(
        operational_fields, on="planet_key", how="left", validate="one_to_one"
    )
    primary_work["selected_primary_extreme"] = (
        primary_work["factorial_width"].ge(config.extreme_width_threshold)
        | primary_work["complete_row_width"].ge(config.extreme_width_threshold)
    )
    primary_work["selected_rank_displacement"] = primary_work["rank_displacement"].astype(float).ge(config.extreme_rank_displacement)
    primary_work["selected_operational_candidate"] = primary_work["top_k_status"].isin(["guaranteed", "possible"])
    stable_keys = set(primary_work.nsmallest(min(10, len(primary_work)), "factorial_width")["planet_key"])
    primary_work["selected_stable_reference"] = primary_work["planet_key"].isin(stable_keys)
    primary_work["selected_secondary_extreme"] = False

    secondary_work = secondary.copy()
    secondary_work["selected_primary_extreme"] = False
    secondary_work["selected_rank_displacement"] = False
    secondary_work["selected_operational_candidate"] = False
    secondary_work["selected_stable_reference"] = False
    secondary_work["selected_secondary_extreme"] = secondary_work["factorial_width"].ge(config.extreme_width_threshold)

    flag_columns = [
        "selected_primary_extreme", "selected_rank_displacement", "selected_operational_candidate",
        "selected_secondary_extreme", "selected_stable_reference",
    ]
    combined = pd.concat([primary_work, secondary_work], ignore_index=True, sort=False)
    combined = combined.loc[combined[flag_columns].any(axis=1)].copy()
    reason_names = {
        "selected_primary_extreme": "primary_extreme",
        "selected_rank_displacement": "rank_displacement",
        "selected_operational_candidate": "operational_candidate",
        "selected_secondary_extreme": "secondary_extreme",
        "selected_stable_reference": "stable_reference",
    }
    combined["selection_reasons"] = combined.apply(
        lambda row: ";".join(reason for column, reason in reason_names.items() if bool(row[column])), axis=1
    )
    combined["review_priority"] = np.select(
        [
            combined["selected_operational_candidate"]
            | combined["factorial_width"].ge(10)
            | combined["rank_displacement"].astype(float).ge(500),
            combined["selected_primary_extreme"] | combined["selected_secondary_extreme"] | combined["selected_rank_displacement"],
            combined["selected_stable_reference"],
        ],
        [0, 1, 3], default=2,
    ).astype(int)
    combined["review_classification"] = "pending"
    combined["review_finding"] = ""
    combined["evidence_reference"] = ""
    combined["reviewer"] = ""
    combined["review_date"] = ""
    combined["second_reviewer"] = ""
    combined["adjudication"] = ""
    return stable_sort(combined, ["review_priority", "factorial_width", "planet_key"])


def load_forensic_review_decisions(path: Path) -> pd.DataFrame:
    """load_forensic_review_decisions()."""
    columns = [
        "planet_key", "observable", "review_classification", "review_finding", "evidence_reference",
        "reviewer", "review_date", "second_reviewer", "adjudication",
    ]
    if not path.is_file():
        return pd.DataFrame(columns=columns)
    decisions = pd.read_csv(path, dtype="string").fillna("")
    missing = set(columns) - set(decisions.columns)
    if missing:
        raise ValueError(f"Forensic review decisions are missing columns: {sorted(missing)}")
    if decisions.duplicated(["planet_key", "observable"]).any():
        raise ValueError("Forensic review decisions must be unique by planet and observable.")
    allowed = {"", "pending", "confirmed_source_difference", "confirmed_agreement", "corrected_source_error", "source_semantic_mismatch", "matching_error", "unit_error", "parameter_semantics_difference", "data_error", "unresolved"}
    invalid = set(decisions["review_classification"].dropna()) - allowed
    if invalid:
        raise ValueError(f"Unknown forensic review classifications: {sorted(invalid)}")
    return decisions[columns].copy()


def apply_forensic_review_decisions(cases: pd.DataFrame, decisions: pd.DataFrame) -> pd.DataFrame:
    """apply_forensic_review_decisions()."""
    if decisions.empty:
        return cases.copy()
    case_keys = set(map(tuple, cases[["planet_key", "observable"]].astype("string").to_numpy()))
    decision_keys = set(map(tuple, decisions[["planet_key", "observable"]].astype("string").to_numpy()))
    unknown = decision_keys - case_keys
    if unknown:
        raise ValueError(f"Forensic review decisions reference unselected cases: {sorted(unknown)[:10]}")
    review_columns = [column for column in decisions.columns if column not in {"planet_key", "observable"}]
    output = cases.drop(columns=review_columns, errors="ignore").merge(
        decisions, on=["planet_key", "observable"], how="left", validate="one_to_one"
    )
    for column in review_columns:
        output[column] = output[column].fillna("")
    output["review_classification"] = output["review_classification"].replace("", "pending")
    return output


def validation_issues_frame(reports: Sequence[ValidationReport]) -> pd.DataFrame:
    """validation_issues_frame()."""
    rows: list[dict[str, Any]] = []
    for report in reports:
        if not report.issues:
            rows.append({"stage": report.stage, "severity": "information", "code": "passed", "message": "No issues.", "count": 0})
        for issue in report.issues:
            rows.append({
                "stage": report.stage, "severity": issue.severity, "code": issue.code,
                "message": issue.message, "count": issue.count,
                "context": json.dumps(dict(issue.context), sort_keys=True, default=json_default),
            })
    return pd.DataFrame(rows)
```

```python
def run_physics_reference_tests(tolerance: float) -> ValidationReport:
    """run_physics_reference_tests()."""
    report = ValidationReport("physics_reference_tests")
    cases = {
        "geometric_area_ratio_ppm": (R_EARTH_OVER_R_SUN ** 2 * 1e6, 83.863467),
        "density_g_cm3": (EARTH_DENSITY_G_CM3, 5.514),
        "gravity_gearth": (1.0, 1.0),
        "escape_velocity_km_s": (EARTH_ESCAPE_VELOCITY_KM_S, 11.186),
        "insolation_searth": (T_SUN_K ** -4 * T_SUN_K ** 4, 1.0),
    }
    for name, (calculated, expected) in cases.items():
        if not math.isclose(calculated, expected, rel_tol=max(tolerance, 1e-7), abs_tol=1e-9):
            report.add("blocking", "reference_value", f"Known-value test failed for {name}.", calculated=calculated, expected=expected)
    report.metrics["tests"] = len(cases)
    return report


def run_ranking_reference_tests() -> ValidationReport:
    """run_ranking_reference_tests()."""
    report = ValidationReport("ranking_reference_tests")
    point = pd.DataFrame({
        "planet_key": ["a", "b", "c"],
        "factorial_lower": [3.0, 2.0, 1.0],
        "factorial_upper": [3.0, 2.0, 1.0],
    })
    point_result = robust_top_k(point, 1)
    status = dict(zip(point_result["planet_key"], point_result["top_k_status"]))
    if status != {"a": "guaranteed", "b": "excluded", "c": "excluded"}:
        report.add("blocking", "point_ranking", "Point-interval top-1 reference test failed.", status=status)
    overlap = pd.DataFrame({
        "planet_key": ["a", "b"],
        "factorial_lower": [1.0, 1.5],
        "factorial_upper": [3.0, 2.5],
    })
    overlap_result = robust_top_k(overlap, 1)
    if not overlap_result["top_k_status"].eq("possible").all():
        report.add("blocking", "overlap_ranking", "Overlapping-interval reference test failed.")
    tied = pd.DataFrame({
        "planet_key": ["a", "b"],
        "factorial_lower": [2.0, 2.0],
        "factorial_upper": [2.0, 2.0],
    })
    tied_result = robust_top_k(tied, 1)
    if not tied_result["top_k_status"].eq("guaranteed").all():
        report.add("blocking", "tie_policy", "Exact-tie minimum-rank policy failed.")
    report.metrics["tests"] = 3
    return report


def run_contract_reference_tests(tolerance: float) -> ValidationReport:
    """run_contract_reference_tests()."""
    report = ValidationReport("data_contract_reference_tests")
    tests = 0

    identifiers = ["  Kepler–10   b  ", "TOI - 123 b", "Kepler-16 (AB) b"]
    for value in identifiers:
        normalized = normalize_identifier(value)
        tests += 1
        if normalize_identifier(normalized) != normalized:
            report.add("blocking", "normalization_not_idempotent", "Identifier normalization is not idempotent.", value=value)
    tests += 1
    if extract_planet_component("Kepler-10 b") != "b":
        report.add("blocking", "component_not_preserved", "Planet component extraction failed.")

    conversion_cases = [
        (1.0, R_JUPITER_TO_R_EARTH),
        (1.0, M_JUPITER_TO_M_EARTH),
    ]
    for original, factor in conversion_cases:
        tests += 1
        if not math.isclose((original * factor) / factor, original, rel_tol=tolerance, abs_tol=tolerance):
            report.add("blocking", "conversion_roundtrip", "A declared unit conversion failed round-trip validation.", factor=factor)

    tests += 1
    if detection_rule("Eclipse Timing Variations").transit_evidence != "unresolved":
        report.add("blocking", "eclipse_timing_rule", "Eclipse timing was incorrectly promoted to confirmed transit evidence.")
    tests += 1
    if detection_rule("Transit").transit_evidence != "confirmed":
        report.add("blocking", "transit_rule", "Explicit transit evidence was not classified as confirmed.")
    tests += 1
    try:
        detection_rule("Unreviewed Synthetic Discovery Method")
        report.add("blocking", "unknown_detection_rule", "An unknown detection label did not fail closed.")
    except ValueError:
        pass

    observable = OBSERVABLES["geometric_area_ratio_ppm"]
    equal = pd.DataFrame({
        "planet_radius_rearth_a": [1.0], "planet_radius_rearth_b": [1.0],
        "star_radius_rsun_a": [1.0], "star_radius_rsun_b": [1.0],
    })
    equal_bounds_a = exact_factorial_bounds(equal, observable)
    equal_bounds_b = exact_factorial_bounds(equal, observable)
    tests += 1
    if not math.isclose(float(equal_bounds_a.loc[0, "factorial_width"]), 1.0, rel_tol=tolerance, abs_tol=tolerance):
        report.add("blocking", "zero_width_identity", "Identical sources did not produce unit width.")
    tests += 1
    if not equal_bounds_a.equals(equal_bounds_b):
        report.add("blocking", "deterministic_repeat", "Repeated deterministic calculation changed its output.")

    perturbed = equal.copy()
    perturbed["planet_radius_rearth_b"] = 2.0
    perturbed_bounds = exact_factorial_bounds(perturbed, observable)
    tests += 1
    if not math.isclose(float(perturbed_bounds.loc[0, "factorial_width"]), 4.0, rel_tol=tolerance, abs_tol=tolerance):
        report.add("blocking", "single_coordinate_width", "Single-coordinate exponent propagation failed.")

    report.metrics["tests"] = tests
    return report


def run_bound_property_tests(trials: int, seed: int, tolerance: float) -> ValidationReport:
    """run_bound_property_tests()."""
    report = ValidationReport("randomized_bound_property_tests")
    generator = np.random.default_rng(seed)
    failures = 0
    for trial in range(trials):
        n_inputs = int(generator.integers(1, 7))
        inputs = tuple(f"x{index}" for index in range(n_inputs))
        exponents = tuple(generator.uniform(-5, 5, size=n_inputs))
        constant = float(np.exp(generator.uniform(-3, 3)))
        observable = PhysicalObservable(
            f"property_{trial}", "property test", "arbitrary", inputs, exponents, constant,
            "test", "randomized identity test", "not a scientific observable",
        )
        values_a = np.exp(generator.uniform(-12, 12, size=n_inputs))
        values_b = np.exp(generator.uniform(-12, 12, size=n_inputs))
        row = {f"{name}_a": [a] for name, a in zip(inputs, values_a)}
        row.update({f"{name}_b": [b] for name, b in zip(inputs, values_b)})
        frame = pd.DataFrame(row)
        bounds = exact_factorial_bounds(frame, observable).iloc[0]
        vertices = enumerate_vertices(frame.iloc[0], observable)
        if not np.allclose(
            [bounds["factorial_lower"], bounds["factorial_upper"]],
            [vertices.min(), vertices.max()], rtol=tolerance, atol=0,
        ):
            failures += 1
    if failures:
        report.add("blocking", "random_bound_failure", "Randomized exact-bound property test failed.", failures)
    report.metrics.update(trials=trials, failures=failures)
    return report


def run_ranking_property_tests(trials: int, seed: int) -> ValidationReport:
    """run_ranking_property_tests()."""
    report = ValidationReport("randomized_ranking_property_tests")
    generator = np.random.default_rng(seed)
    failures = 0
    for trial in range(trials):
        n = int(generator.integers(2, 8))
        lower = np.exp(generator.uniform(-4, 4, size=n))
        upper = lower * np.exp(generator.uniform(0.01, 2, size=n))
        frame = pd.DataFrame({
            "planet_key": [f"p{index}" for index in range(n)],
            "factorial_lower": lower,
            "factorial_upper": upper,
        })
        calculated = robust_top_k(frame, max(1, n // 2)).set_index("planet_key")
        all_ranks = {key: [] for key in frame["planet_key"]}
        for choices in itertools.product((0, 1), repeat=n):
            values = np.where(np.asarray(choices, dtype=bool), upper, lower)
            ranks = pd.Series(values).rank(method="min", ascending=False).astype(int).to_numpy()
            for key, rank in zip(frame["planet_key"], ranks):
                all_ranks[key].append(int(rank))
        for key, ranks in all_ranks.items():
            if int(calculated.loc[key, "best_rank"]) != min(ranks) or int(calculated.loc[key, "worst_rank"]) != max(ranks):
                failures += 1
                break
    if failures:
        report.add("blocking", "random_ranking_failure", "Randomized endpoint-enumeration ranking test failed.", failures)
    report.metrics.update(trials=trials, failures=failures)
    return report


physics_reference_report = run_physics_reference_tests(CONFIG.floating_tolerance)
ranking_reference_report = run_ranking_reference_tests()
contract_reference_report = run_contract_reference_tests(CONFIG.floating_tolerance)
bound_property_report = run_bound_property_tests(CONFIG.bound_property_trials, CONFIG.seed + 101, CONFIG.floating_tolerance)
ranking_property_report = run_ranking_property_tests(CONFIG.ranking_property_trials, CONFIG.seed + 202)
display(pd.concat([
    report_table(physics_reference_report), report_table(ranking_reference_report),
    report_table(contract_reference_report),
    report_table(bound_property_report), report_table(ranking_property_report),
], ignore_index=True))
if not all(report.passed for report in (physics_reference_report, ranking_reference_report, contract_reference_report, bound_property_report, ranking_property_report)):
    raise RuntimeError("Reference tests failed before empirical analysis.")
```

## Conservative matching

Candidates are generated by normalized planet identifier. Period agreement is displayed for
all candidates before tier assignment. Tier 1 requires a unique identifier in each source and
period consistency within $0.1\%$, where the relative period discrepancy is
$$
\delta_P = \frac{|P_A - P_B|}{\min(P_A, P_B)}.
$$
This is a matching rule, not independent validation evidence.


All tie handling uses minimum-rank (competition) semantics: when multiple rows share an identical score they receive the same rank, and the next distinct score receives a rank equal to the number of rows already assigned.  A top-$k$ set defined by minimum rank may therefore contain more than $k$ members, because all rows tied at the boundary rank are included.

## Limitations and prohibited interpretations

1. Two derivative source artifacts do not span all epistemic or measurement uncertainty.
2. Coordinate-wise hybrid vertices need not correspond to a published complete parameter set.
3. Shared literature means the sources are not independent replications.
4. Complete-case cohorts are selected toward better-characterized systems.
5. The geometric area ratio is not an observed transit depth or full target-priority metric.
6. Mass-dependent results are secondary because mass semantics may differ.
7. Host resampling measures composition sensitivity, not uncertainty about the frozen estimand.
8. Extreme cases require source-provenance interpretation and are not automatically errors.
9. The analysis measures reproducibility under substitution, not truth, causality, or habitability.
10. Observed discrepancies confound catalog curation decisions with the temporal information-epoch difference between sources (Source A extends to ~February 2018; Source B is frozen in March 2026); disentangling these two effects requires access to same-epoch catalog versions.
11. Of the 444 selected forensic cases, only all 113 priority-0 cases received documented self-audit review; the remaining 331 lower-priority cases have not been manually reviewed and their classification is provisional.
